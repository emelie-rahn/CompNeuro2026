```python
import matplotlib.pyplot as plt
import numpy as np
```

Homework 2, problem 1, part c

```python
a = np.linspace(0.01, 0.99, 400)
c = (1 - 2*a) / np.sqrt(a * (1 - a)) # D = 1

plt.figure(figsize = (6,4))
plt.plot(a, c, 'b-', label=r'$c(a) = \frac{1-2a}{\sqrt{a(1-a)}}$') # velocity curve

plt.axhline(2, linestyle='--', label='Passive Speed (c = 2)')
plt.axhline(0, linestyle='-')


plt.axvline(0.5, linestyle=':', label='Stalling Point (a = 0.5)')

# intersection point where c = 2
cross_a = (2 - np.sqrt(2)) / 4
plt.plot(cross_a, 2, 'ko')

# Layout and labels setup
plt.xlim(0, 1)
plt.ylim(-5, 5)
plt.xlabel('Excitation Threshold (a)')
plt.ylabel('Velocity (c)')
plt.title('Velocity vs Threshold')

plt.legend()
plt.show()
```
